```python
import pandas as pd
```

```python
df = pd.read_csv('/rt_reviews.csv', encoding='latin-1')
```

```python
df_cleaned = df.drop_duplicates()
df_subset = df_cleaned.sample(n=3000, random_state=2026).reset_index(drop=True)
df_subset.index = df_subset.index + 1
display(df_subset)
```

```python
df_subset.to_csv('rt_reviews_subset.csv', index=True, index_label='Row_ID')
print("Subset saved successfully as 'rt_reviews_subset.csv'!")
```
